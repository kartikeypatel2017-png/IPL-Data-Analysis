# 02 - Merge Matches & Deliveries

Joins the cleaned match-level data with ball-by-ball delivery data into one flat table, and checks the result for missing values.

**Input:** `data/processed/matches_c.csv`, `data/processed/deliveries_c.csv`
**Output:** `Merge_table` (in-memory, used for further analysis)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
matches_c_df = pd.read_csv('../data/processed/matches_c.csv')
```

```python
deliveries_c_df= pd.read_csv('../data/processed/deliveries_c.csv')
```

```python
Merge_table = pd.merge(matches_c_df, deliveries_c_df, left_on='id', right_on='match_id', how='left')
```

```python
Merge_table.isnull().sum()
```
